# Indian Road Accident Severity Prediction

This project follows the KDD (Knowledge Discovery in Databases) process and uses a simple predictive modelling approach to classify accident severity.

## Objective
- Understand the dataset
- Clean and prepare the data
- Train a predictive model
- Evaluate the model using standard metrics


## Step 1: Data Collection

The dataset contains accident records collected across multiple Indian cities and includes traffic, environmental, and road-condition variables.


```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Step 1: Load the dataset
df = pd.read_csv("indian_roads_dataset.csv")
print("Dataset Shape:", df.shape)
print(df.head())
print("\nMissing Values:\n", df.isnull().sum())
```

## Step 2: Data Cleaning and Preprocessing

The dataset is checked for missing values and transformed into a format suitable for model training.


```python
features = [
    "hour",
    "day_of_week",
    "is_weekend",
    "road_type",
    "lanes",
    "traffic_signal",
    "weather",
    "visibility",
    "temperature",
    "traffic_density",
    "is_peak_hour",
    "festival"
]

target = "accident_severity"

X = df[features]
y = df[target]

categorical_cols = [
    "day_of_week",
    "road_type",
    "weather",
    "visibility",
    "traffic_density",
    "festival"
]

X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
X = X.fillna(X.median(numeric_only=True))
X = X.fillna(0)

encoder = LabelEncoder()
y = encoder.fit_transform(y)

print("\nTarget Classes:")
for i, label in enumerate(encoder.classes_):
    print(i, "=", label)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

## Step 3: Model Building

A K-Nearest Neighbors classifier is trained to predict accident severity using the prepared dataset.


```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

y_pred = knn.predict(X_test_scaled)
```

## Step 4: Model Evaluation

The trained model is evaluated using accuracy, classification report, and confusion matrix.


```python
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", round(accuracy, 4))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred, target_names=encoder.classes_))
print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))
```

## Conclusion

This project follows the KDD and predictive modelling stages in a simple and structured way. The dataset was collected, cleaned, transformed, modelled, and evaluated to understand accident severity patterns using road and traffic-related factors.
